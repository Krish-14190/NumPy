```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
print("view shape of given matrix : " , sales_data.shape)
print("\n 1st three restaurant data : " , sales_data[:3])
```

```python
#total sales per year : np.sum:
sale_year_2021 = np.sum(sales_data[:,1])
print("total sales of 2021 year is : " , sale_year_2021)
sale_year_column = np.sum(sales_data[:,1:],axis=0)#column wise
print("total sale per year is :\n ", sale_year_column)
sale_year_rows = np.sum(sales_data[:,1:],axis=1)#year wise
print("row wise sale per year :\n " , sale_year_rows)


#minimum sale per year :
min_sale = np.min(sales_data[:,1:],axis=1)
print("minimum sale per year is : \n " , min_sale)


#maximum sale per year :
max_sale = np.max(sales_data[:,1:],axis=1)
print("maximum sale per year is : \n " , max_sale)


#maximum sale per year among all restaurant :
max_sale_all = np.max(sales_data[:,1:],axis=0)
print("maximum sale per year among all restaurant : \n " , max_sale_all)

#average sale per restraurant :
avg_sale = np.mean(sales_data[:,1:],axis=1)
print("avearge sale per restraurant : \n ", avg_sale)
```
