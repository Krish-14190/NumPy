### NUMPY ARRAY AND BASICS:

```python
import numpy as np
```

```python
# this is a 1-dimensional array :
arr_1D = np.array([1,2,3,4,5])
print("1D array : " , arr_1D)

# this is a 2-dimensional array :
arr_2D = np.array([[1,2,3],[4,5,6]])
print("2D array: \n" , arr_2D)
```

### list vs numpy :

```python
py_list = [1,2,3]
print(py_list*2)

np_array = np.array([1,2,3])
print(np_array*2)
```

```python
import time
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("operation time : " ,time.time()-start)


start =time.time()
np_array = np.arange(1000000) *2
print("operation time : " , time.time()-start)
```

### creating array from scratch

```python
zeros = np.zeros((3,4))
print("zeros array : \n" , zeros)

ones = np.ones((2,3))
print("ones array : \n" , ones)

full = np.full((3,3) , 5)
print("full array : \n" , full)

random = np.random.random((2,3))
print("random array : \n" , random)

sequence = np.arange(0,11,2)
print("sequence array : \n" , sequence)
```

 ### vector , matrix and tensor

```python
vector = np.array([1,2,3])
print("Vector : " , vector)

matrix = np.array([[1,2,3],
                   [4,5,6]])
print("Matrix : " , matrix)

tensor = np.array([[[1,2],[3,4]],
                    [[5,6],[7,8]]])
print("Tensor : " , tensor)
tensor_1= np.array([[[3,4],[5,6]],
                    [[7,8],[9,10]]])
print("tensor_1 : " , tensor_1)
```

### Array Properties :

```python
arr = np.array([[1,2,3],
                [4,5,6]])
print("Shape : " , arr.shape)
print("Dimension :" , arr.ndim)
print("Size : " , arr.size)
print("DType : " , arr.dtype)
```

### Reshaping Array :

```python
arr = np.arange(12)
print("Original array : \n " , arr)

reshaped = arr.reshape(3,4)
print("Reshaped array : \n " , reshaped)

flattened =reshaped.flatten()
print("Flattened array : \n " , flattened)

#original array :
raveled = reshaped.ravel()
print("Raveled array : \n " , raveled)

transpose = reshaped.T
print("Transpose array : \n " , transpose)
```
